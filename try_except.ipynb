# Python `try` and `except`

Use `try` and `except` to handle errors gracefully. Python runs the code in `try`; if a matching error occurs, it runs the corresponding `except` block instead of stopping the program.

## Handling invalid number input

Converting text that is not a number with `int()` raises a `ValueError`. Catch it to show a helpful message.

```python
user_input = "hello"

try:
    number = int(user_input)
    print("The number is", number)
except ValueError:
    print("Please provide text that can be converted to an integer.")
```

## Handling division by zero

Dividing a number by zero raises a `ZeroDivisionError`. Catch that specific exception.

```python
numerator = 10
denominator = 0

try:
    result = numerator / denominator
    print("Result:", result)
except ZeroDivisionError:
    print("A number cannot be divided by zero.")
```

## Using `else` and `finally`

`else` runs only when the `try` block has no error. `finally` runs whether an error happened or not.

```python
try:
    number = int("25")
except ValueError:
    print("That was not a valid integer.")
else:
    print("Conversion succeeded:", number)
finally:
    print("This message always runs.")
```

## Reusable functions with exceptions

Functions can validate their arguments and raise a clear exception. The caller can then decide how to handle it. This example also uses `*numbers`, which lets a function accept any number of positional arguments.

```python
def calculate_average(*numbers: float) -> float:
    """Return the average of one or more numbers."""
    if not numbers:
        raise ValueError("At least one number is required.")
    return sum(numbers) / len(numbers)


try:
    average = calculate_average(8, 12, 16)
    print("Average:", average)
except ValueError as error:
    print("Could not calculate average:", error)

try:
    calculate_average()
except ValueError as error:
    print("Could not calculate average:", error)
```
