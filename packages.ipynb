# Python Packages

A **module** is a Python file that contains reusable code. A **package** groups related modules together so they can be organized and imported. Python's standard library includes many packages, and you can also install packages written by others.

## Importing from a package

`urllib` is a package in Python's standard library. The `parse` module inside it provides helpers for working with URLs.

````python
from urllib.parse import urlparse, urlencode

address = urlparse("https://python.org/learn")
print(address.netloc)

query = urlencode({"name": "Sam", "topic": "Python"})
print(query)
````

## Import styles

You can import a module from a package, or import the package module and use its full name. An alias can make a long module name shorter.

````python
import urllib.parse
import urllib.parse as url_tools

print(urllib.parse.urlparse("https://example.com").netloc)
print(url_tools.urlparse("https://example.com").netloc)
````

## Making your own package

A simple package can be a folder containing Python modules. For example:

```text
my_tools/
    __init__.py
    greetings.py
```

If `greetings.py` defines `say_hello`, you can import it from another Python file like this:

```python
from my_tools.greetings import say_hello

say_hello("Ava")
```

The `__init__.py` file is commonly used to mark a folder as a package and can be empty. Modern Python also supports namespace packages without it.

## Installing third-party packages

`pip` installs packages from the Python Package Index (PyPI). In a notebook, use `%pip` so the package is installed in the notebook's Python environment. Replace `package-name` with the package you need; this example is intentionally not run.

````python
# %pip install package-name
````

After installing a package, import its module in Python using `import module_name`. The package name used by `pip` and the name used in an `import` statement can sometimes be different.
