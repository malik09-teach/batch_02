# Python Packages

A **module** is a Python file that contains reusable code. A **package** groups related modules together so they can be organized and imported. Python's standard library includes many packages, and you can also install packages written by others.

## Importing from a package

`urllib` is a package in Python's standard library. The `parse` module inside it provides helpers for working with URLs.

```python
from urllib.parse import urlparse, urlencode

address = urlparse("https://python.org/learn")
print(address.netloc)

query = urlencode({"name": "Sam", "topic": "Python"})
print(query)
```

## Import styles

You can import a module from a package, or import the package module and use its full name. An alias can make a long module name shorter.

```python
import urllib.parse
import urllib.parse as url_tools

print(urllib.parse.urlparse("https://example.com").netloc)
print(url_tools.urlparse("https://example.com").netloc)
```

## Creating and importing our own package

This project contains a package folder named `my_package`. Its `greetings.py` module defines a `greet` function, and `__init__.py` makes that function available directly from the package. Run this notebook from the project folder to import it.

```python
from my_package import greet

message = greet("Ava")
print(message)
```

The function can also be imported directly from its module with `from my_package.greetings import greet`. Importing it from `my_package` works because `__init__.py` re-exports `greet`.
