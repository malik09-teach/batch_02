```python
#3*3

rows=3
cols=3
current_num=1

matrix=[]

for i in range(rows):
    
    cur_row=[]
    
    for j in range(cols):
        
        cur_row.append(current_num*2)
        current_num+=1

    matrix.append(cur_row)
```

```python
matrix
```

```python
for row in matrix:
    for item in row :
        print(item,end=" ")
    print()    
```

```python
[item for row in matrix for item in row]
```
