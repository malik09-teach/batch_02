```python
#** 

def info(**person_info):
    print("keys and values.................")
    for k,v in person_info.items():
        print(f"{k}:{v}")
    
    print("keys .................")
    for k in person_info.keys():
        print(f"{k}")
        
    print("values.................")    
    for v in person_info.values():
        print(f"{v}")     
        
        
        
    print("dic.................")    
    print(person_info)
```

```python
info(name="iqra",sem="3rd",age="19")
```

```python
def config(model_name,*layers,**hyper_parameters):
    
    print(f"model name:{model_name}")
    
    
    for layer  in layers:
        print("layer type:",layer)
        
    for parm,ofsetval  in hyper_parameters.items():
        print(f"{parm}:{ofsetval}")
        
    if hyper_parameters["verbose"] ==True:
        loss=0.15
        for i in range(100):
            loss+=0.3
            print("loss",loss)   
        
    
```

```python
config("ANN","input_layer","dense_layer(8,128,64,32,8)","output_layer",learning_rate=0.01,epochs=10,verbose=True)
```

```python
def math(num1,num2):
    num11=num1
    num22=num2
    
    def add(num11,num22):
        print("add",num11+num22)


    def sub(num11,num22):
        print("sub",num11-num22)

        
    def mul(num11,num22):
        
        print("mul",num11*num22)  
        num111=num11
        num222=num22
        
        
        def power(num111,num222):
            print("power:",num111**num222)
        
        power(num111,num222)
        
        
    add(num11,num22) 
    sub(num11,num22)
    mul(num11,num22)
         
```

```python
math(2,7)
```
