```python
def max_num(x):
    maxx_num=x[0]
    for i in x:
        if i >maxx_num:
            maxx_num=i
    return  maxx_num
        
        
        
        
```

```python
def security(func):
    def warper():
        print("system is under checking......")
        func()
        print ("action complete....")
    return warper    
        
```

```python
@security
def delet_rec():
    print("all records delete.")
    
    
@security
def update_dec():
    print("update rec")
        
    
```

```python
delet_rec()
```

```python
update_dec()
```
