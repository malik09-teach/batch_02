```python
list1=[0,4,5,1,2,8,3,88]
```

```python
def max_num(x):
    maxx_num=x[0]
    for i in x:
        if i >maxx_num:
            maxx_num=i
    return  maxx_num
        
        
        
        
```

```python
maximum=max_num(list1)
```

```python
print("maximum number is :",maximum)
```
