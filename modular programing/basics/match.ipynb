```python
cmd=input("enter cmd")

match cmd:
    case "quit" | "exit":
        print("system log out")

    case  "load wps":
        print(f"loading file wps")
    case _:
         print("no match ")
```

```python
num1=int(input("enter num1:"))
opreation=str(input("enter (+,-,*,/,//,%):"))
num2=int(input("enter num2:"))


match opreation:
    case "+":
        print(num1+num2)
    case "-":
        print(num1-num2)
    case "*":
        print(num1*num2)
    case "/":
        print(num1/num2)         
    case "//":
        print(num1//num2)
    case "%":
        print(num1%num2)
    case _:
        print("invalid operator....")        








```
